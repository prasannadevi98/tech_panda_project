```python
import pandas as pd
```

```python
data=pd.read_csv(r"C:\Users\acer\Downloads\Car_Sales_CSV\heart-disease.csv")
print(data)
```

```python
data=data[["sex","oldpeak"]]

df=data.head(50)
print(df)
```

```python
frequency={}

for sex, oldpeak in zip(df["sex"],df["oldpeak"]):
    key=(sex,oldpeak)
    if key in frequency:
        frequency[key]+=1
    else:
        frequency[key] =1 

print(frequency)          
```

```python
#for my reference
largest_value=0
largest_key=0
for key,value in zip(frequency.keys(),frequency .values()):
        if value>largest_value:
                largest_value=value
                largest_key=key
print(largest_value,largest_key)      
```

```python
#convert frequency dic into csv

import pandas as pd

df_csv = pd.DataFrame(list(frequency.items()), columns=["sex_oldpeak", "count"])
df_csv.to_csv("sex_oldpeak_count.csv",index=False)
df_csv
```

```python
import matplotlib.pyplot as plt

plt.bar(df_csv["sex_oldpeak"].astype(str), df_csv["count"])
plt.xticks(rotation=90)
plt.xlabel('sex_oldpeak')
plt.ylabel('count')
plt.show()
```
